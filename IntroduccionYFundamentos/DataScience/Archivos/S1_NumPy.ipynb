## Bloque 0 · Qué es data science

Data science es el trabajo de **responder preguntas con datos para tomar mejores decisiones**. No es una herramienta ni un algoritmo concreto: es un proceso que se repite en cualquier proyecto, sea en un banco, en un hospital o en el servicio de bicis de una ciudad. El machine learning es una parte de ese proceso, la que construye modelos para predecir, pero no la única.

### Tres disciplinas que se cruzan

| Disciplina | Para qué |

## Bloque 4 · Broadcasting

Para operar arrays de formas distintas, NumPy compara las dimensiones **de derecha a izquierda**. Son compatibles si son iguales o si una vale 1.

| A | B | Resultado |
|---|---|---|
| (3, 4) | (4,) | (3, 4) |
| (3, 4) | (3, 1) | (3, 4) |
| (3, 4) | (3,) | error |

### Demo

````python
grid = np.array([[2, 8, 5, 1], [0, 3, 9, 4], [6, 6, 2, 0]])
print(grid * np.array([0.5, 1, 1, 0.5]))        # a row (4,) applies to every row

totals = grid.sum(axis=1)
try:
    grid / totals                                  # (3, 4) / (3,) fails
except ValueError as err:
    print("Error:", err)
print((grid / grid.sum(axis=1, keepdims=True)).round(2))   # (3, 4) / (3, 1) works
````
